```python
import matplotlib as plt
import pandas as pd
```

```python
ibm = pd.read_csv("data/IBM.csv")
```

```python
ibm.plot()
```

```python
ibm.plot(y="Close")
```

```python
def rank_performance(stock_price):
    if stock_price <=50:
        return "Poor"
    elif 50<=stock_price<=100:
        return "Satisfactory"
    else:
        return "Excellent"

ibm["Close"].apply(rank_performance).value_counts().plot(kind="bar")
```

```python

ibm["Close"].apply(rank_performance).value_counts().plot(kind="pie",legend=True)
```
